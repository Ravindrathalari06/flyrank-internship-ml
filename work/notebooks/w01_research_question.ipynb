## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
from datasets import load_dataset
import pandas as pd

# Load starter dataset


dataset = load_dataset("FlyRank/internship-starter")

df = dataset["train"].to_pandas()

# Basic dataset size
total_content = len(df)

# Recent performance coverage
recent_data_count = df["impressions_last_30d"].notna().sum()

# Descriptive trend count
down_count = (df["trend_direction"] == "down").sum()
down_percentage = down_count / total_content * 100

print(f"Total content items: {total_content:,}")
print(f"Content items with recent impression data: {recent_data_count:,}")
print(f"Content items currently marked as down: {down_count:,} ({down_percentage:.2f}%)")
```
